## 6. Time Series Visualization

```python
# Plot single simulation run
def plot_simulation_run(sim_run=1, fault_num=1, n_sensors=6):
    """
    Plot sensor readings for a single simulation run
    """
    # Filter data
    sim_data = df[(df['simulationRun'] == sim_run) & 
                  (df['faultNumber'] == fault_num)].copy()
    
    if sim_data.empty:
        print(f"No data for simulation {sim_run}, fault {fault_num}")
        return
    
    # Select sensors to plot
    sensors = xmeas_cols[:n_sensors]
    
    # Create subplots
    fig, axes = plt.subplots(n_sensors, 1, figsize=(15, 2*n_sensors))
    
    for i, sensor in enumerate(sensors):
        ax = axes[i] if n_sensors > 1 else axes
        ax.plot(sim_data['sample'], sim_data[sensor], linewidth=0.8)
        ax.set_ylabel(sensor)
        ax.grid(alpha=0.3)
        
        # Mark anomaly region (fault introduced at hour 1)
        fault_start = 20  # 1 hour = 20 samples (3 min intervals)
        ax.axvline(x=fault_start, color='red', linestyle='--', 
                   alpha=0.5, label='Fault introduced')
        
        if i == 0:
            ax.legend()
            ax.set_title(f'Simulation Run {sim_run}, Fault Type {fault_num}',
                        fontsize=14, fontweight='bold')
    
    axes[-1].set_xlabel('Sample Number')
    plt.tight_layout()
    plt.show()

# Example: Plot first simulation with fault 1
plot_simulation_run(sim_run=1, fault_num=1, n_sensors=6)
```

## 7. Normal vs Fault Comparison

```python
# Compare normal (fault 0) vs faulty operation
def compare_normal_vs_fault(sensor='xmeas_1', fault_type=1, n_samples=500):
    """
    Compare sensor behavior in normal vs fault conditions
    """
    # Get first simulation of each type
    normal = df[(df['faultNumber'] == 0) & 
                (df['simulationRun'] == df[df['faultNumber']==0]['simulationRun'].iloc[0])].head(n_samples)
    faulty = df[(df['faultNumber'] == fault_type) & 
                (df['simulationRun'] == df[df['faultNumber']==fault_type]['simulationRun'].iloc[0])].head(n_samples)
    
    fig, axes = plt.subplots(2, 2, figsize=(15, 8))
    
    # Time series comparison
    axes[0, 0].plot(normal['sample'], normal[sensor], label='Normal', alpha=0.7)
    axes[0, 0].plot(faulty['sample'], faulty[sensor], label=f'Fault {fault_type}', alpha=0.7)
    axes[0, 0].set_title(f'{sensor} - Time Series Comparison')
    axes[0, 0].set_xlabel('Sample')
    axes[0, 0].set_ylabel('Value')
    axes[0, 0].legend()
    axes[0, 0].grid(alpha=0.3)
    
    # Distribution comparison
    axes[0, 1].hist(normal[sensor], bins=50, alpha=0.6, label='Normal', density=True)
    axes[0, 1].hist(faulty[sensor], bins=50, alpha=0.6, label=f'Fault {fault_type}', density=True)
    axes[0, 1].set_title(f'{sensor} - Distribution Comparison')
    axes[0, 1].set_xlabel('Value')
    axes[0, 1].set_ylabel('Density')
    axes[0, 1].legend()
    axes[0, 1].grid(alpha=0.3)
    
    # Box plots
    data_to_plot = [normal[sensor].dropna(), faulty[sensor].dropna()]
    axes[1, 0].boxplot(data_to_plot, labels=['Normal', f'Fault {fault_type}'])
    axes[1, 0].set_title(f'{sensor} - Box Plot Comparison')
    axes[1, 0].set_ylabel('Value')
    axes[1, 0].grid(alpha=0.3)
    
    # Statistics table
    stats_data = {
        'Metric': ['Mean', 'Std', 'Min', 'Max', 'Median'],
        'Normal': [
            normal[sensor].mean(),
            normal[sensor].std(),
            normal[sensor].min(),
            normal[sensor].max(),
            normal[sensor].median()
        ],
        f'Fault {fault_type}': [
            faulty[sensor].mean(),
            faulty[sensor].std(),
            faulty[sensor].min(),
            faulty[sensor].max(),
            faulty[sensor].median()
        ]
    }
    stats_df = pd.DataFrame(stats_data)
    axes[1, 1].axis('tight')
    axes[1, 1].axis('off')
    table = axes[1, 1].table(cellText=stats_df.values,
                            colLabels=stats_df.columns,
                            cellLoc='center',
                            loc='center')
    table.auto_set_font_size(False)
    table.set_fontsize(10)
    table.scale(1, 2)
    axes[1, 1].set_title('Statistical Comparison')
    
    plt.tight_layout()
    plt.show()

# Example comparison
compare_normal_vs_fault(sensor='xmeas_7', fault_type=1, n_samples=500)
```

## 8. Correlation Analysis

```python
# Correlation matrix for subset of sensors (memory efficient)
# Use first 10,000 samples for correlation analysis
sample_data = df.head(10000)

# Select subset of sensors
sensors_subset = xmeas_cols[:15]  # First 15 measured variables

# Calculate correlation
corr_matrix = sample_data[sensors_subset].corr()

# Plot heatmap
plt.figure(figsize=(12, 10))
sns.heatmap(corr_matrix, annot=True, fmt='.2f', cmap='coolwarm', 
            center=0, square=True, linewidths=1)
plt.title('Sensor Correlation Matrix (xmeas 1-15)', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()

# Find highly correlated pairs
print("\nHighly Correlated Sensor Pairs (|r| > 0.8):")
print("="*50)
for i in range(len(corr_matrix.columns)):
    for j in range(i+1, len(corr_matrix.columns)):
        if abs(corr_matrix.iloc[i, j]) > 0.8:
            print(f"{corr_matrix.columns[i]} <-> {corr_matrix.columns[j]}: {corr_matrix.iloc[i, j]:.3f}")
```

## 9. Data Preparation for ML / IndustryFlow

```python
# Prepare data for ML training
def prepare_ml_data(fault_types=[0, 1, 2, 3], max_samples_per_fault=10000):
    """
    Prepare balanced dataset for ML training
    """
    ml_data = []
    
    for fault in fault_types:
        fault_data = df[df['faultNumber'] == fault].head(max_samples_per_fault)
        ml_data.append(fault_data)
    
    ml_df = pd.concat(ml_data, ignore_index=True)
    
    print(f"ML Dataset Prepared:")
    print(f"  Total samples: {len(ml_df):,}")
    print(f"  Fault distribution:")
    print(ml_df['faultNumber'].value_counts().sort_index())
    print(f"  Memory: {ml_df.memory_usage(deep=True).sum() / 1024**2:.2f} MB")
    
    return ml_df

# Create ML dataset
ml_dataset = prepare_ml_data(fault_types=[0, 1, 2, 3, 4, 5], max_samples_per_fault=5000)
```

```python
# Feature and target separation
X = ml_dataset[xmeas_cols + xmv_cols].values
y = ml_dataset['faultNumber'].values

print(f"Features (X): {X.shape}")
print(f"Target (y): {y.shape}")
print(f"\nClass distribution:")
unique, counts = np.unique(y, return_counts=True)
for u, c in zip(unique, counts):
    print(f"  Fault {u}: {c:,} samples")
```

## 11. IndustryFlow Integration Example

```python
# Format data for IndustryFlow ingestion
def format_for_industryflow(sim_run=1, fault_num=1, max_samples=100):
    """
    Convert TEP data to IndustryFlow format:
    - tenant_id
    - equipment_id  
    - timestamp
    - sensor readings
    - anomaly label
    """
    # Get simulation data
    sim_data = df[(df['simulationRun'] == sim_run) & 
                  (df['faultNumber'] == fault_num)].head(max_samples).copy()
    
    # Add IndustryFlow fields
    sim_data['tenant_id'] = 'tep_demo'
    sim_data['equipment_id'] = f'reactor_{sim_run}'
    
    # Create timestamps (3 minute intervals)
    start_time = datetime.now()
    sim_data['timestamp'] = [start_time + timedelta(minutes=3*i) 
                             for i in range(len(sim_data))]
    
    # Anomaly label (1 hour after start)
    fault_start_sample = 20  # 1 hour = 20 samples
    sim_data['is_anomaly'] = (sim_data['sample'] >= fault_start_sample).astype(int)
    
    # Reorder columns
    cols_order = ['tenant_id', 'equipment_id', 'timestamp', 'faultNumber', 
                  'is_anomaly', 'simulationRun', 'sample'] + xmeas_cols + xmv_cols
    
    return sim_data[cols_order]

# Example conversion
industryflow_data = format_for_industryflow(sim_run=1, fault_num=1, max_samples=50)
print("IndustryFlow Format Sample:")
print(industryflow_data[['tenant_id', 'equipment_id', 'timestamp', 'faultNumber', 
                          'is_anomaly', 'xmeas_1', 'xmeas_2', 'xmeas_3']].head(10))
```

```python
# Export IndustryFlow format
industryflow_data.to_csv(OUTPUT_PATH + 'TEP_IndustryFlow_Format.csv', index=False)
print(f"✓ Exported: TEP_IndustryFlow_Format.csv")
print(f"  Shape: {industryflow_data.shape}")
print(f"  Ready for IndustryFlow ingestion!")
```
